```python
import urllib.request
import zipfile
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# 1. Download and load the standard SMS Spam Collection dataset
url = "https://archive.ics.uci.edu/static/public/228/sms+spam+collection.zip"
urllib.request.urlretrieve(url, "sms.zip")

with zipfile.ZipFile("sms.zip", "r") as zip_ref:
    zip_ref.extractall("sms_data")

df = pd.read_csv("sms_data/SMSSpamCollection", sep="\t", names=["label", "message"])

# 2. Encode labels (ham = 0, spam = 1)
df["target"] = df["label"].map({"ham": 0, "spam": 1})

# 3. Train-Test Split (80/20)
X_train, X_test, y_train, y_test = train_test_split(
    df["message"], df["target"], test_size=0.2, random_state=42, stratify=df["target"]
)

# 4. Feature Extraction using TF-IDF (removing English stopwords)
tfidf = TfidfVectorizer(stop_words="english", lowercase=True, max_features=3000)
X_train_vec = tfidf.fit_transform(X_train)
X_test_vec = tfidf.transform(X_test)

# 5. Train Multinomial Naive Bayes
nb_model = MultinomialNB()
nb_model.fit(X_train_vec, y_train)
y_pred_nb = nb_model.predict(X_test_vec)

print("--- Multinomial Naive Bayes Results ---")
print(f"Accuracy: {accuracy_score(y_test, y_pred_nb):.4f}")
print(classification_report(y_test, y_pred_nb, target_names=["Ham", "Spam"]))

# 6. Train Logistic Regression
lr_model = LogisticRegression(max_iter=1000)
lr_model.fit(X_train_vec, y_train)
y_pred_lr = lr_model.predict(X_test_vec)

print("--- Logistic Regression Results ---")
print(f"Accuracy: {accuracy_score(y_test, y_pred_lr):.4f}")
print(classification_report(y_test, y_pred_lr, target_names=["Ham", "Spam"]))
```
