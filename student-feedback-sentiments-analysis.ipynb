```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix

# 1. Representative Student Feedback Dataset (Positive, Negative, Neutral)
data = {
    "feedback": [
        "The professor explains difficult concepts very clearly and with practical examples.",
        "Course material was poorly structured and deadlines were unreasonable.",
        "Lectures were average, covering basic topics without much depth.",
        "Excellent practical assignments that helped me understand machine learning.",
        "Worst course experience, teaching assistants were completely unhelpful.",
        "The pacing was okay, standard theoretical slides with standard exercises.",
        "Inspiring instructor, always available to answer doubts after class.",
        "Outdated curriculum with obsolete tools, not relevant to current industry.",
        "Decent introduction to the topic, neither exceptionally good nor bad.",
        "Loved the real-world case studies and hands-on lab sessions.",
        "Evaluation and grading were extremely biased and delayed for weeks.",
        "The syllabus was completed on time, standard textbook content throughout.",
        "The instructor was engaging and made every complex topic simple.",
        "Lacked practical coding examples, purely theoretical and boring.",
        "Fair evaluation, covered all promised topics adequately.",
        "Outstanding guidance on the final semester capstone project.",
        "Poor audio quality in recordings and disorganized reading materials.",
        "Normal lecture delivery, slides provided after class as expected."
    ] * 25,  # Replicate samples to simulate a 450-record dataset
    "sentiment": [
        "Positive", "Negative", "Neutral",
        "Positive", "Negative", "Neutral",
        "Positive", "Negative", "Neutral",
        "Positive", "Negative", "Neutral",
        "Positive", "Negative", "Neutral",
        "Positive", "Negative", "Neutral"
    ] * 25
}

df = pd.DataFrame(data)

# 2. Train-Test Split (80/20) with stratification
X_train, X_test, y_train, y_test = train_test_split(
    df["feedback"], df["sentiment"], test_size=0.2, random_state=42, stratify=df["sentiment"]
)

# 3. Text Preprocessing & TF-IDF Vectorization
# Removing standard English stopwords, lowercasing, and using unigrams + bigrams
tfidf = TfidfVectorizer(stop_words="english", lowercase=True, ngram_range=(1, 2), max_features=1500)
X_train_vec = tfidf.fit_transform(X_train)
X_test_vec = tfidf.transform(X_test)

# 4. Train Multiclass Logistic Regression
model = LogisticRegression(max_iter=1000, multi_class="multinomial")
model.fit(X_train_vec, y_train)

# 5. Evaluate Performance
y_pred = model.predict(X_test_vec)

print("--- Student Feedback Sentiment Analysis Results ---")
print(f"Overall Accuracy: {accuracy_score(y_test, y_pred) * 100:.2f}%\n")
print("Classification Report:")
print(classification_report(y_test, y_pred))

# 6. Test on Unseen Feedback
sample_reviews = [
    "The teacher gave clear explanations and helpful feedback.",
    "Very bad management, labs were canceled without notice.",
    "Topics were covered as outlined in the syllabus."
]
sample_vec = tfidf.transform(sample_reviews)
predictions = model.predict(sample_vec)

for text, pred in zip(sample_reviews, predictions):
    print(f"[{pred}] -> {text}")
```
